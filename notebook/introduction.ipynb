```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
from example_repo import load, bar_graph
```

I will investigate cash/card payments by category off clothing categories. The dataset can be explored using bar charts that I will create using MatPlotLib. I hope to build a prediction model off this data. Research question: Is there a distinct association between an item bought and the payment method used (cash/card) in this dataset? 
Hypothesis: Yes, there exists a difference between categories.

```python
df = load()
df
```

```python
# Payment Percentages by Item Purchased
# pd.crosstab() calculates the percentage of each payment method for each item purchased.
# Two dataframes: df["Item Purchased"] and df["Payment Method"] are the two columns used to create the crosstab.
payment_percentages = pd.crosstab(
    df["Item Purchased"],
    df["Payment Method"],
    normalize="index"
) * 100
# payment_percentages.round(1) rounds the percentages to one decimal place for better readability.
payment_percentages.round(1)
```

The above code uses pandas and creates a DataFrame. It looks at the columns "Item Purchased" and "Payment Method" in the dataset and creates a chart. The chart looks at items in the dataset and the payment methods use (card/cash). The payment methods are divided into percentages for each category of item.

```python
# payment_percentages is the pandas dataframe created earlier
# payment_percentages.plot() creates a bar chart to visualize the payment percentages by item purchased 
ax = payment_percentages.plot(
    kind="bar",
    # stacked=True puts the payment methods on top of each other in the bar chart, making it easier to compare the total percentages for each item purchased.
    stacked=True,
    figsize=(10, 6)
)

ax.set_title("Payment Percentages by Item Purchased")
ax.set_xlabel("Item Purchased")
ax.set_ylabel("Percentage of purchases")
# set_ylim(0, 100) sets the y-axis limits to be between 0 and 100, which is reasonable for the given dataset
ax.set_ylim(0, 100)

# Rotate the x-axis labels for better readability
plt.xticks(rotation=45, ha="right")
# Add a legend to the chart to indicate which color corresponds to which payment method
plt.legend(title="Payment Method")
# tight_layout() adjusts the layout of the chart to make sure that all elements fit within the figure area and are not cut off.
plt.tight_layout()
plt.show()
```

The above graph compares the frequency of cash and card payments across different item categories. I would like to investigate the pattern across different categories. For example, flip flops are more frequently bought with cash whereas trousers are more frequently bought by card.
